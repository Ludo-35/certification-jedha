CLEANING :

Tout d'abord, on supprime les ponctuations gênantes, on transforme en minuscules :

FIRST SPLITTING :

Nous allons procéder plus bas à un undersampling... pour tenter de compenser la sous-représentativité des SPAMs vis-à-vis des HAMs. Voilà pourquoi nous allons effectuer deux splits successifs afin de garder de côté un set de validation, qui ne sera pas under-samplé, pour garder une proportion de spams plus réaliste au final :

```python
data_train_test, data_val = train_test_split(data, test_size=0.15, stratify=data['v1'], random_state=2)
```

UNDERSAMPLING :

Notre target est donc assez déséquilibrée. L'undersampling, tout comme un oversampling, va peut-être permettre une amélioration des performances au moment de l'apprentissage.

Définition de la fonction d'under-sampling.
On définie un nombre de fois (nb_reduce) où les non-spams seront présents par rapport au nombre de spams.
On prend tous les spams, on les compte, et on prend ensuite au hasard nb_spams X nb_reduce parmi les non-spams. Puis ensuite, on concatène les deux parties et on re-mélange :

```python
def data_under_sampling(data, nb_reduce):
  data_sample_2 = data.loc[data['v1']=='spam',:]
  len_class_sub = len(data_sample_2)
  data_sample_1 = data.loc[data['v1']=='ham',:].sample(nb_reduce * len_class_sub)
  
  return pd.concat([data_sample_1,data_sample_2], axis=0).sample(frac=1).reset_index(drop=True)
```

Après plusieurs essais de valeurs entières pour nb_reduce, partant de 1 (équilibre total entre les spams et les hams), il semble qu'un bon équilibre se trouve avec la valeur 4 (4 fois plus de hams que de spams), au niveau des performances sur le validation set. (car si on se focalise seulement sur le train set et le test set, c'est la valeur 1 qui est nécéssairement la meilleure valeur, voilà pourquoi il était important de constituer le validation set).

```python
data_red = data_under_sampling(data_train_test, 4)
```

```python
data_red['v1'].value_counts()
```

PADDING :

Ici, on retient le token le plus long, pour servir de base au padding :

```python
max_len_sentence = max([len(token) for token in data['v2_tokenized'].to_list()])
```

On applique le padding, pour uniformiser la longueur des tokens, avec des valeurs nulles en fin de phrase :

```python
v2_pad_train = tf.keras.preprocessing.sequence.pad_sequences(data_red['v2_tokenized'], padding="post", maxlen=max_len_sentence)
v2_pad_val = tf.keras.preprocessing.sequence.pad_sequences(data_val['v2_tokenized'], padding="post", maxlen=max_len_sentence)
```

PREPROCESSING DE LA TARGET :

On utilise un simple Label Encoder pour convertir "ham" et "spam" en O et 1...

```python
preprocessor_y = LabelEncoder()
```

```python
y_train_test = preprocessor_y.fit_transform(data_red['v1'])
y_val = preprocessor_y.transform(data_val['v1'])
```

SECOND SPLITTING :

Après avoir effectué les bons preprocessings sur le set (train + test) undersamplé, d'une part, et le set de validation, d'autres part... on peut procéder au second splitting pour constituer les sets de train et de test. On utilise un stratifier pour conserver les proportions de la target dans les deux sets, comme lors du premier splitting :

```python
x_train, x_test, y_train, y_test = train_test_split(v2_pad_train, y_train_test, test_size=0.15, stratify=y_train_test, random_state=2)
```

Conversion en datasets batchés (avec un batchsize de 128, là encore plusieurs essais ont permis de trouver ce qui semble être une bonne valeure). Pour le train set, on effectue également un shuffle comme il se doit :

```python
BATCH_SIZE = 64
```

```python
train_batch = tf.data.Dataset.from_tensor_slices((x_train, y_train)).shuffle(len(x_train)).batch(BATCH_SIZE)
test_batch = tf.data.Dataset.from_tensor_slices((x_test, y_test)).batch(BATCH_SIZE)

print('Preprocessing done --- extract of test_batch : ')
for sentence, target in test_batch.take(1):
  print(sentence, target)
```

Prélecture pour optimiser les temps de calculs lors de l'entraînement :

```python
AUTOTUNE = tf.data.AUTOTUNE

train_batch = train_batch.cache().prefetch(buffer_size=AUTOTUNE)
test_batch = train_batch.cache().prefetch(buffer_size=AUTOTUNE)
```

### 4/ Premier Modèle avec Embeder + Pooling + Couches Denses

ARCHITECTURE :

Nous constituons ici un premier modèle simple pour répondre à la problématique, et tenter d'obtenir un modèle "baseline". Pas de couches de RNN pour l'instant... Il n'y a qu'une couche d'embedding en entrée, qui accueille les tokens un à un, sur chaque phrase... puis le Global Average Pooling vient aggréger l'ensemble des valeurs de tous les tokens de la phrase, ne tenant ainsi pas en compte l'aspect séquentiel de la phrase. Ensuite, les différentes valeurs des différents neurones de l'Embedding (au nombre de 128) sont envoyées vers trois couches succesives de couches de neurones denses avec fonction d'activation simple Rectified Linear Unit ("relu")... Enfin, le dernier neurone de la dernière couche possède une fonction d'activation sigmoid, dédiée  la classification binaire classique.

```python
model_1 = tf.keras.Sequential([        
                  tf.keras.layers.Embedding(vocab_size+1, 128, input_shape=[x_train.shape[1],],name="embedding"),
                  tf.keras.layers.GlobalAveragePooling1D(),
                  tf.keras.layers.Dense(64, activation='relu'),
                  tf.keras.layers.Dropout(0.2),
                  tf.keras.layers.Dense(32, activation='relu'),
                  tf.keras.layers.Dense(8, activation='relu'),
                  tf.keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model_1.summary()
```

COMPILATION DU MODÈLE ET MISE EN PLACE DU 'TRACKING' DE PERFORMANCES :

Définition de l'optimiseur, la loss, et les fonctions de métriques, qui sont compilés pour le modèle. 

Comme nous allons utiliser pour chaque modèle les même paramètres, nous créons ici une fonction pour éviter la redondance.
Par ailleurs, dans cette même fonction, nous allons permettre au modèle de se pré-"logger" pour le "tracking" des performances avec Tensorboard, qui est un bon outil de suivi de la loss fonction et des métriques, surtout quand il s'agit de comparer des modèles.

Optimiseur : on utilise Adam qui permet de faire décroitre le learning rate au fur-et-à-mesure de la descente de gradient.
Pour la fonction de coût (loss function), on prend la Binary Cross Entropy, adaptée aux cas de classification binaire.
Et enfin, trois métriques adaptés à la classification:
- l'accuracy, pour la performance globale
- la précision : combien de spams vrais et positifs par rapport à tous les positifs ?
- le 'recall' (ou sensibilité) : combien de spams faux et négatifs par rapport à tous les véritables spams ?

#### À propos de la métrique "précision"

La métrique "précision" est ici très importante, c'est une métrique qui est sans doute à minimiser. En effet, il vaut sans doute mieux laisser passer des spams en ham plutôt que l'inverse (on compte sur l'utilisateur pour faire un second tri... en revanche, il risque d'être en colère s'il n'a pas reçu un SMS important, parce que nous l'avons par erreur classé en spam).

```python
def compile_and_log(model, model_name='default', architecture={}):
  # definition de l'optimiseur - on utilise Adam()
  optimizer= tf.keras.optimizers.Adam()
  # lors de la compilation, pour la loss on prend
  model.compile(optimizer="adam",
              loss=tf.keras.losses.BinaryCrossentropy(),
              metrics = [tf.keras.metrics.BinaryAccuracy(name='accuracy'),
                         tf.keras.metrics.Precision(name='precision'),
                         tf.keras.metrics.Recall(name='recall')])
  log_dir="logs/fit/" + model_name + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
  tensorboard_callback = tf.keras.callbacks.TensorBoard(log_dir=log_dir, histogram_freq=1)
  hparams_callback = hp.KerasCallback(log_dir, architecture)
  print (f"{model} logged with name : {model_name}...")
  return tensorboard_callback, hparams_callback
  
```

```python
tensorboard_callback, hparams_callback = compile_and_log(model_1, "Simple Multi-Layers Denses", {
                                                                                            'dim_embedding':128,
                                                                                            'nb_dense_units':105,
                                                                                            'dropout':0.2})
```

TRAINING :

```python
model_1.fit(
    train_batch,
    epochs=20,
    verbose=1,
    callbacks=[tensorboard_callback, hparams_callback],
    validation_data=test_batch)
```

ANALYSE :

Nous avons d'emblée de très bons résultats (proches de 1 sur toutes les métriques, à la fois sur le train set et le test set)... Par ailleurs, nous pouvons observer la courbe de loss sur tensorboard.

NOTE : Pour l'ensemble des modèles, tensorboard est activé en toute fin de ce notebook (il faut relancer la cellule à chaque entraînement d'un nouveau modèle) pour pouvoir comparer les performances de l'ensemble des modèles entrainés.

La courbe de loss converge bien, sans instabilité ni overfitting apparent.

De bonnes performances avec ce premier modèle de base, OK ... Mais attention : il faut maintenant vérifier sur le validation set, qui a été épargné par l'undersampling.

Pour cela, il faut définir des fonctions (qui seront réutilisées sur chaque modèle) qui permettent l'analyse de ces performances sur le validation set.

La première fonction permet de retourner les résultats-prédictions du model sur le val set (float de 0 à 1) en valeure entière, pour pouvoir être comparée aux valeurs réelles de la target. Le treshold, limite de répartition des 2 classes 0 et 1, par défaut initialisé à 0.5 pour une répartition équitable, peut éventuellement être changé pour pouvoir ajuster la répartition des faux positifs et des faux négatifs, améliorant ainsi soit la sensibilité soit le treshold.
La deuxième fonction sert quant à elle à afficher les résultats sous forme de matrice de de confusion, et d'afficher également les métriques sur le validation set.

```python
def predictions_singleton_to_binary(model, tresh, x_val, y_val):
  y_true = y_val
  y_pred = np.where(model.predict(x_val) > tresh, 1, 0)
  return y_true, y_pred
```

```python
def display_matrix_and_scores(y_true, y_pred):
  ConfusionMatrixDisplay.from_predictions(y_true, y_pred)
  print(f'Accuracy : {accuracy_score(y_true,y_pred)}')
  print(f'Precision : {precision_score(y_true,y_pred)}')
  print(f'Recall : {recall_score(y_true,y_pred)}')
```

```python
y_true, y_pred = predictions_singleton_to_binary(model_1, 0.5, v2_pad_val, y_val)
```

```python
display_matrix_and_scores(y_true, y_pred)
```

Nous avons d'assez belles performances avec une accuracy de 0,987 et une précision de 0,96... ce qui est le plus important ! Le 'recall' est quant à lui plus bas (0,94)...

Exemple d'utilisation du treshold si on veut encore plus maximiser la précision des prédictions. Ici, on monte le treshold à 0,9, ce qui veut dire qu'un résultat probabiliste en sortie de modèle de 85% de chances d'être un spam, par exemple, va classer le message en non-spam tout de même. 

```python
y_true, y_pred = predictions_singleton_to_binary(model_1, 0.9, v2_pad_val, y_val)
```

```python
display_matrix_and_scores(y_true, y_pred)
```

### 5 / Des modèles simples de type Recurrent Neural Network (RNN)

Passons maintenant à des modèles toujours construits par nous-même, mais avec des couches de RNN...
Ici, nous commençons par un montage plutôt traditionnel... Il y a toujours l'Embedding, avec le même nombre de dimensions qu'auparavant (128). Ensuite, nos couches de RNNs qui vont assimiler les tokens 1 à 1, en gardant en mémoire la trace des tokens précédents, pour conserver une "mémoire" de la phrase.
La deuxième et dernière couche de RNN retourne uniquement le résultat en sortie des tokens (et non la séquence entière) pour le transmettre à une couche de layers Dense de 8 neurones, et enfin notre unique neurone charger de la distinction des 2 classes, comme dans le premier modèle.

Comme pour le premier modèle, nous avons essayé nombre de configurations possibles (nombres de couches, nombres de neurones dans chaque couches...). Egalement, nous avons essayé l'emploi de RNNs plus complexes (GRU et LSTM)... Ces derniers se sont soldés par une énorme instabilité et/ou d'incroyables échecs (précision à 0, le modèle semble faire de l'underfitting)... Il semblerait que les modèles trop complexes sont incompatibles avec le type de phrases des SMS, peut-être assez chaotiques et peu construits par nature ? (ce n'est qu'une hypothèse).
Nous présentons ici uniquement les deux modèles avec des simples RNN...

```python
model_2 = tf.keras.Sequential([        
                  tf.keras.layers.Embedding(vocab_size+1, 128, input_shape=[x_train.shape[1],],name="embedding"),
                  tf.keras.layers.SimpleRNN(units=64, return_sequences=True),
                  tf.keras.layers.SimpleRNN(units=32, return_sequences=False),
                  tf.keras.layers.Dropout(0.2),
                  tf.keras.layers.Dense(8, activation='relu'),
                  tf.keras.layers.Dropout(0.1),
                  tf.keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model_2.summary()
```

```python
tensorboard_callback, hparams_callback = compile_and_log(model_2, "Simple Multi-Layers RNN with final Dense", {
                                                                                            'dim_embedding':128,
                                                                                            'nb_rnn_units':96,
                                                                                            'nb_units_dense':9,
                                                                                            'dropout':0.1})
```

```python
model_2.fit(
    train_batch,
    epochs=20,
    verbose=1,
    callbacks=[tensorboard_callback, hparams_callback],
    validation_data=test_batch)
```

```python
y_true, y_pred = predictions_singleton_to_binary(model_2, 0.5, v2_pad_val, y_val)
```

```python
display_matrix_and_scores(y_true, y_pred)
```

Les performances de ce modèle avec RNNs tiennent à peu près la route (accuracy de 0,95) mais elles sont nettement inférieures à notre premier modèle, notamment le "recall" est très mauvais...
Par ailleurs, quand on observe la fonction côut (loss) sur tensorboard, on prend connaissance d'une très grande instabilité, et d'une remontée spectaculaire à partir de l'epoch 12.. Il faudrait augmenter éventuellement la taille du batch, mais même en le faisant, après essais, l'instabilité règne toujours. On pourrait également arrêter l'entrainement avant la remontée. Mais de toutes les façons, le modèle semble nettement inférieur au premier...

Nous testons ensuite une deuxième série de modèles avec une conception un peu particulière :
Il n'y a pas de couche de Denses en fin de chaîne (mis à part le neurone final de classification binaire)... mais à la place un Global Average Pooling qui "applatit toute la séquence" retournée préalablement par le réseau de layers Denses. L'intérêt ici est de tenter de conserver jusqu'au bout, et le plus tard possible, la "mémoire" captée par les RNNs, avec une moyenne effectuée sur la séquence entière. C'est un simple essai.

```python
model_3 = tf.keras.Sequential([        
                  tf.keras.layers.Embedding(vocab_size+1, 128, input_shape=[x_train.shape[1],],name="embedding"),
                  tf.keras.layers.SimpleRNN(units=64, return_sequences=True),
                  tf.keras.layers.SimpleRNN(units=32, return_sequences=True),
                  tf.keras.layers.SimpleRNN(units=16, return_sequences=True),
                  tf.keras.layers.GlobalAveragePooling1D(),
                  tf.keras.layers.Dropout(0.1),
                  tf.keras.layers.Dense(1, activation="sigmoid")
])
```

```python
tensorboard_callback, hparams_callback = compile_and_log(model_3, "Simple Multi-Layers RNN with Global Average Pooling", {
                                                                                            'dim_embedding':128,
                                                                                            'nb_rnn_units':112,
                                                                                            'nb_units_dense':1,
                                                                                            'dropout':0.1})
```

```python
model_3.fit(
    train_batch,
    epochs=20,
    verbose=1,
    callbacks=[tensorboard_callback, hparams_callback],
    validation_data=test_batch)
```

```python
y_true, y_pred = predictions_singleton_to_binary(model_3, 0.5, v2_pad_val, y_val)
```

```python
display_matrix_and_scores(y_true, y_pred)
```

Les performances sont visiblement un peu meilleures par rapport au modèle précédent. 
Mais quand on regarde la courbe de loss sur tensorflow, il y a une incroyable instabilité, avec des remontées, des corrections, etc... 
Donc le modèle, dans l'abosolu, est assez équivalent au précédent.

### 6 / Modèles importés pré-entrainés

Nous allons ici tenter d'importer deux modèles pré-entrainés.

Le premier est un simple modèle distilBERT (version distillée de BERT [lien vers Hugging Face ici](https://huggingface.co/docs/transformers/model_doc/distilbert) ) qui peut être utilisé pour de la classification textuelle. Il a l'avantage de conserver les propriétés de BERT, et presque toutes les performances, mais avec plus de rapidité et d'efficacité, avec un nombre de layers divisé par deux. DistilBERT a été créé grâce à une "distillation" de Bert, dont le principe repose sur un entrainement conjoint "maître-élève" (voir le principe expliqué ici : https://medium.com/huggingface/distilbert-8cf3380435b5)

```python
model = TFAutoModel.from_pretrained('distilbert-base-uncased')
```

Le tokenizer associé :

```python
tokenizer = AutoTokenizer.from_pretrained('distilbert-base-uncased')
```

PREPROCESSING :

Il nous faut revenir un peu en arrière lors de notre préprocessing (que nous avions utilisé avec les modèles faits à la main), car le modèle possède son propre tokenizer spécial, autant l'utiliser. De plus, c'est un tokenizer qui encode des informations supplémentaires (comme l' 'attention_mask', le masque d'attention) dont le modèle aura besoin pour fonctionner.
Faut-il revenir au texte lemmatisé, au texte cleané ?... Non, il nous faut garder le texte tel qu'il était au tout départ, car justement le tokenizer DistilBert est un tokenizer entrainé qui prend en compte la ponctuation en insérant des tokens spéciaux pour cela... mieux vaut ne pas intervenir.

```python
x_train, x_test, y_train, y_test = train_test_split(data_red['v2'], y_train_test, test_size=0.2, stratify=y_train_test)
x_val = data_val['v2']
```

On définit deux fonctions pour le tokenizer spécial.

La première prend chaque groupe de tokens (chaque SMS) et renvoie la tokenisation correspondante dans le bon format pour le modèle.

On prend une max_lenght un peu élargi par rapport à la max_lenght constatée précédemment, car il faut faire de la place pour les tokens spéciaux du tokenizer. On ajoute 20 pour être large.

```python
def tokenize_sentence(sentence):
    tokens = tokenizer.encode_plus(sentence, max_length=max_len_sentence + 20,
                                   truncation=True, padding='max_length',
                                   add_special_tokens=True, return_attention_mask=True,
                                   return_token_type_ids=False, return_tensors='tf')
    return tokens['input_ids'], tokens['attention_mask']
```

Exemple avec un SMS :

```python
tokenize_sentence('Hello, I want to contact you. Cause... u win a Ipad !')
```

On constate que le tokenizer a généré plus de tokens dans la phrase (19, ou 18 si on enlève le "start") que le nombre de mots de départs (17 en comptant la ponctuation). C'est le principe de ce tokenizer qui ajoute des tokens informationnels au milieu de la séquence comme les ponctuations, les débuts de phrases, etc... grâce aux tokens aditionnels suivants :


```python
tokenizer.all_special_tokens
```

La deuxième fonction va permettre de convertir en une fois tout un dataset, en prenant chaque séquence (chaque message) un par un, et en stokant les résultats dans deux matrices numpys retournées, une pour la séquence, une pour le masque d'attention.

```python
def tokenize_series(df):
  Xids = np.zeros((len(df), max_len_sentence+20))
  Xmask = np.zeros((len(df), max_len_sentence+20))
  for i, sentence in enumerate(df):
    Xids[i, :], Xmask[i, :] =  tokenize_sentence(sentence)
  return Xids, Xmask
```

On tokenize nos trois datasets :

```python
X_ids_train, X_mask_train = tokenize_series(x_train)
X_ids_test, X_mask_test = tokenize_series(x_test)
X_ids_val, X_mask_val = tokenize_series(x_val)
```

Et on convertit les formats numpy array en datasets avec à chaque fois, nos deux inputs (la séquence de tokens "ids" et la séquence de "mask") et notre target. Pour cela, on peut employer une méthode qui consiste à entrer un dictionnaire comme input dans le tensor slice, afin de référencer directement les deux inputs et leurs names dans le dataset ainsi constitué.

```python
train_batch = tf.data.Dataset.from_tensor_slices(({'input_ids': X_ids_train, 'attention_mask' : X_mask_train} , y_train)).shuffle(len(x_train)).batch(BATCH_SIZE)
test_batch = tf.data.Dataset.from_tensor_slices(({'input_ids': X_ids_test, 'attention_mask' : X_mask_test} , y_test)).batch(BATCH_SIZE)
```

Il reste à permettre la pré-lecture de nos datasets, pour accélérer les calculs.

```python
AUTOTUNE = tf.data.AUTOTUNE

train_batch = train_batch.cache().prefetch(buffer_size=AUTOTUNE)
test_batch = train_batch.cache().prefetch(buffer_size=AUTOTUNE)
```

ARCHITECTURE :

Nous ne pouvons utiliser un modèle Sequential avec une double entrée dans le modèle à constituer.
Nous définissons donc:
- deux couches d'Inputs (une pour chaque entrée 'input_ids' et 'attentionmask') nécéssaires au fonctionnement de DistilBERT
- l'embeddings, qui n'est autre que le modèle DistilBERT importé
- puis un GlobalMaxPooling et une Normalisation, qui vont permettre de lisser (en moyenne) la sortie de DistilBert...
- une simple couche de layers Dense de 32 unités
- un Dropout pour l'overfitting
- et enfin notre unique neurone final, comme à l'accoutumée

```python
input_ids = tf.keras.layers.Input(shape=(max_len_sentence + 20,), name='input_ids', dtype='int32')
mask = tf.keras.layers.Input(shape=(max_len_sentence + 20,), name='attention_mask', dtype='int32')

embeddings = model(input_ids, attention_mask=mask)[0] # pour prendre seulement le "last hidden state" du modèle

l = tf.keras.layers.GlobalMaxPool1D()(embeddings)
l = tf.keras.layers.BatchNormalization()(l)
l = tf.keras.layers.Dense(32, activation='relu')(l)
l = tf.keras.layers.Dropout(0.1)(l)

outputs = tf.keras.layers.Dense(1, activation='sigmoid', name='outputs')(l)

model_4 = tf.keras.Model(inputs=[input_ids, mask], outputs=outputs)
```

```python
model_4.layers[2].trainable = False
```

### 7 / Tentative d'ajouts d'informations en Pipeline parallèles (Multi-Inputs)

Une dernière tentative. Nous cherchons encore à améliorer le modèle pour le rendre parfait (est-ce possible ?)...
Nous nous intéressons ici à la longueur des SMS et également la fréquence de leur caractère numérique.
Un SPAM est-il en moyenne plus long (ou plus court) qu'un non-SPAM ?
Un SPAM comporte-t-il plus ou moins de caractères numériques ? 
Pourrait-on alors ajouter ces informations cruciales dans le modèle, par des pipelines parallèles ?...

Regardons tout d'abord la longueur des SMS, dans le dataset initial, selon s'il sont des spams ou non :

```python
data['len'] = data['v2'].apply(lambda s : len(s))
```

```python
fig = px.histogram(data.loc[data['v1']=='ham',:], x='len', title='Repartition of HAM SMS in function of the lenght')
fig.show()
```

```python
fig = px.histogram(data.loc[data['v1']=='spam',:], x='len', title='Repartition of SPAM SMS in function of the lenght')
fig.show()
```

La longueur semble en effet être un élément important. La plupart des non-spams ont un pic de distribution aux alentours de 30 caractères, alors que les spams ont un pic autour des 155...

Regardons maintenant la distribution selon le nombre de caractères numériques :

```python
data['alpha_count'] = data['v2'].apply(lambda s : len([char for char in s if char.isnumeric()]))
```

```python
fig = px.histogram(data.loc[data['v1']=='ham',:], x='alpha_count', title='Repartition of HAM SMS in function of the numerics characters count', nbins=70)
fig.show()
```

```python
fig = px.histogram(data.loc[data['v1']=='spam',:], x='alpha_count', title='Repartition of SPAM SMS in function of the numerics characters count', nbins=70)
fig.show()
```

Les spams ont une distribution assez chaotiques mais le gros des comptes de charactères numériques se situent aux alentours de 19 charactères / message...
alors que pour les non-spams, la plupart des messages n'ont aucun caractères numériques.

Nous allons faire du "feature engineering" à partir de ces indications, et créer quatre nouvelles features qui vont représenter des distances euclidiennes à ces pics de distributions : 

```python
data['len_d0'] = data['len'].apply(lambda v : np.sqrt((v - 25)**2))
data['len_d1'] = data['len'].apply(lambda v : np.sqrt((v - 155)**2))
data['alpha_d0'] = data['alpha_count']
data['alpha_d1'] = data['alpha_count'].apply(lambda v : np.sqrt((v - 19)**2))
```

Liste des features créees :

```python
add_info_list = ['len_d0', 'len_d1', 'alpha_d0', 'alpha_d1']
```

Il faut reprendre le pré-processing encore plus haut, au moment des splittings, pour pouvoir conserver toutes ces features. Ici, c'est le même processing qu'avant mais revu "en accéléré" :

```python
data_train_test, data_val = train_test_split(data, test_size=0.15, stratify=data['v1'], random_state=2) #premier splitting

data_red = data_under_sampling(data_train_test, 4) #undersampling

preprocessor_y = LabelEncoder() #preprocessing de la target
y_train_test = preprocessor_y.fit_transform(data_red['v1'])
y_val = preprocessor_y.transform(data_val['v1'])

df_train, df_test, y_train, y_test = train_test_split(data_red, y_train_test, test_size=0.15, stratify=y_train_test) #second splitting
df_val = data_val

x_train = df_train['v2'] #constitution des np array pour les messages en inputs du modèle DistilBERT
x_test = df_test['v2']
x_val = df_val['v2']

X_ids_train, X_mask_train = tokenize_series_with_special(x_train) # tokenisation spéciale du modèle et transformation en 2 inputs 
X_ids_test, X_mask_test = tokenize_series_with_special(x_test)
X_ids_val, X_mask_val = tokenize_series_with_special(x_val)
```

Par contre, on ajoute un preprocessing supplémentaire pour les features additionnelles nouvellement crées... On constitue une nouvelle entrée matricielle avec 4 features numériques de distance euclidennes aux "pics" de distribution caractéristiques, en normalisant le tout avec un Standard Scaler :

```python
preprocessor_add_info = StandardScaler()
X_add_info_train = preprocessor_add_info.fit_transform(df_train[add_info_list])
X_add_info_test = preprocessor_add_info.fit_transform(df_test[add_info_list])
X_add_info_val = preprocessor_add_info.fit_transform(df_val[add_info_list])
```

Création des datasets batchés (méthode dictionnaire) :

```python
train_batch = tf.data.Dataset.from_tensor_slices(({'input_ids': X_ids_train, 'attention_mask' : X_mask_train, 'add_info' : X_add_info_train} , y_train)).shuffle(len(x_train)).batch(BATCH_SIZE)
test_batch = tf.data.Dataset.from_tensor_slices(({'input_ids': X_ids_test, 'attention_mask' : X_mask_test, 'add_info' : X_add_info_test} , y_test)).batch(BATCH_SIZE)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_batch = train_batch.cache().prefetch(buffer_size=AUTOTUNE)
test_batch = train_batch.cache().prefetch(buffer_size=AUTOTUNE)
```

On modifie l'architecture pour prendre en compte cette nouvelle entrée des features numériques additionnelles. Pour cela on crée une seconde pipeline parallèle avec seulement deux couches de layers Dense. Une première de 16 neurones, puis une autre de 8...
Ensuite on concatène la sortie du modèle de base DistilBERT (avec une output de dimension 16) avec la sortie de cette nouvelle pipeline d'infos additionnelles (de dimension 8)... Cela laisse plus de "place", toujours, au modèle DistilBert à ce stade.
Puis une nouvelle couche de layers Dense de 8 intermédiaire prend en charge la sortie de la concaténation finale, avant de l'envoyer au neurone de claissification final.

```python
input_ids = tf.keras.layers.Input(shape=(max_len_sentence + 20,), name='input_ids', dtype='int32')
mask = tf.keras.layers.Input(shape=(max_len_sentence + 20,), name='attention_mask', dtype='int32')
add_info = tf.keras.layers.Input(shape=(4,), name='add_info', dtype='int32')

embeddings = model(input_ids, attention_mask=mask)[0]  
l = tf.keras.layers.GlobalMaxPool1D()(embeddings) 
l = tf.keras.layers.BatchNormalization()(l)
l = tf.keras.layers.Dense(32, activation='relu')(l)
l = tf.keras.layers.Dense(16, activation='relu')(l)
output_Dist = tf.keras.layers.Dropout(0.1)(l)

output_add  = tf.keras.layers.Dense(16, "relu")(add_info)
#output_add = tf.keras.layers.Dense(8, "relu")(m)

concatenation = tf.keras.layers.Concatenate(axis=1)([output_Dist, output_add])

x = tf.keras.layers.Dense(8, "relu")(concatenation)

output = tf.keras.layers.Dense(1, 'sigmoid', name='outputs')(x)

model_5_bis = tf.keras.Model(inputs=[input_ids, mask, add_info], outputs=output)
```

```python
model_5_bis.fit(
    train_batch,
    epochs=20,
    verbose=1,
    callbacks=[tensorboard_callback, hparams_callback],
    validation_data=test_batch)
```

RAS. Les résultats ne sont pas altérés à ce stade (du moins, ils ne semblent pas être dégradés). Encore une fois, on pourrait mettre moins d'epochs, ou arrêter l'entrainement plus tôt, mais pour comparer tous les modèles dans le cadre de ce notebook, on préfère garder à chaque fois le même nombre d'epochs.

Notre légère amélioration est-elle le fruit du hasard ? Difficile de conclure à ce stade... Il faudrait revenir en arrière et tenter le même principe de cette pipeline d'informations aditionnelles parallèles avec un modèle plus mauvais pour mieux s'en rendre compte....
