# 🚕 Uber Pick-Ups Clustering — New York City

**Objectif métier :** Uber constate que les chauffeurs ne sont pas toujours positionnés là où la demande se trouve, ce qui allonge le temps d'attente des utilisateurs au-delà du seuil critique de 5-7 minutes.

**Objectif du projet :** construire un algorithme capable d'identifier automatiquement les zones de forte demande ("hot zones") à New York, jour par jour, afin de pouvoir à terme recommander aux chauffeurs où se positionner.

**Démarche :**
1. Exploration et nettoyage des données de courses Uber (avril → septembre 2014)
2. Analyse exploratoire de l'activité par jour de la semaine et par heure
3. Clustering sur un couple (jour, heure) précis pour valider la méthode, avec deux algorithmes : **KMeans** et **DBSCAN**
4. Généralisation du clustering à chaque jour de la semaine (du **lundi** au **dimanche**)
5. Comparaison des deux algorithmes et tableau de bord interactif final


## 1. Imports

```python
import numpy as np
import pandas as pd

import plotly.io as pio
import plotly.express as px
import plotly.graph_objects as go

from sklearn.cluster import KMeans, DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score, davies_bouldin_score

pd.set_option('display.max_columns', None)
pio.renderers.default = 'notebook'
```

## 2. Chargement des données

On dispose de 6 fichiers de courses Uber à New York, un par mois, d'avril à septembre 2014. On les concatène en un seul DataFrame.

*(Le fichier `taxi-zone-lookup.csv` et le jeu de données 2015 (`uber-raw-data-janjune-15`) utilisent un identifiant de zone (`locationID`) au lieu de coordonnées GPS : ils correspondent à un autre exercice sur ce même dataset. On se concentre ici sur les données 2014 en Lat/Lon, qui permettent un clustering géographique direct.)*

```python
import gc

fichiers_2014 = [
    'uber-raw-data-apr14.csv',
    'uber-raw-data-may14.csv',
    'uber-raw-data-jun14.csv',
    'uber-raw-data-jul14.csv',
    'uber-raw-data-aug14.csv',
    'uber-raw-data-sep14.csv',
]

# On ne charge que les colonnes utiles (Lat/Lon en float32) pour rester léger en mémoire
# sur ~4.5 millions de lignes.
morceaux = []
for f in fichiers_2014:
    morceaux.append(pd.read_csv(
        f, usecols=['Date/Time', 'Lat', 'Lon'],
        dtype={'Lat': 'float32', 'Lon': 'float32'},
    ))
df = pd.concat(morceaux, ignore_index=True)
del morceaux
gc.collect()

print(f"{len(df):,} courses chargées".replace(',', ' '))
df.head()
```

## 3. Nettoyage & feature engineering

On convertit `Date/Time` en `datetime` puis on en extrait les variables temporelles utiles pour l'analyse : année, mois, jour, jour de la semaine et heure.

```python
dt = pd.to_datetime(df['Date/Time'])

# Ordre calendaire lundi -> dimanche, utilisé partout dans le notebook
JOURS_ORDRE = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
JOURS_FR = {
    'Monday': 'Lundi', 'Tuesday': 'Mardi', 'Wednesday': 'Mercredi',
    'Thursday': 'Jeudi', 'Friday': 'Vendredi', 'Saturday': 'Samedi', 'Sunday': 'Dimanche',
}

df['day_of_week'] = pd.Categorical(dt.dt.day_name(), categories=JOURS_ORDRE, ordered=True)
df['hour'] = dt.dt.hour.astype('int8')
df['jour_fr'] = df['day_of_week'].map(JOURS_FR)
df['jour_fr'] = pd.Categorical(df['jour_fr'], categories=[JOURS_FR[j] for j in JOURS_ORDRE], ordered=True)

# On ne garde que les colonnes utiles pour la suite (économie mémoire sur 4.5M lignes)
df = df.drop(columns=['Date/Time'])
del dt
gc.collect()

df.head()
```

### 4.2 Activité par jour de la semaine et par heure

On regarde comment l'activité totale (cumulée sur les 6 mois) se répartit selon le jour de la semaine et l'heure, afin d'identifier le pic d'activité qui servira de cas d'étude pour le clustering.

Contrairement à une intuition naïve, ce n'est pas le samedi soir qui concentre le plus de courses : l'activité Uber en semaine, en fin de journée (heure de pointe du soir), est globalement plus dense que le week-end. On retient ce couple (jour, heure) comme cas d'étude pour la suite.

## 5. Clustering sur un jour et une heure précis

Avant de généraliser à toute la semaine, on valide la méthode sur le couple (jour, heure) identifié comme le plus chargé. On y compare **KMeans** et **DBSCAN**.

```python
sous_ensemble = df[(df['day_of_week'] == JOUR_PIC) & (df['hour'] == HEURE_PIC)]
X = sous_ensemble[['Lat', 'Lon']].to_numpy()
print(f"{len(sous_ensemble):,} courses pour {JOURS_FR[JOUR_PIC]} {HEURE_PIC}h".replace(',', ' '))
```

On n'applique pas de standardisation : latitude et longitude sont déjà deux grandeurs continues du même ordre de grandeur (quelques dixièmes de degré sur la zone de NYC), donc les mettre à la même échelle ne changerait pas la géométrie des clusters — et une standardisation classique déformerait l'interprétation géographique directe des centroïdes.

### 5.1 KMeans

```python
wcss, silhouette_kmeans = [], []
k_range = range(2, 11)

for k in k_range:
    km = KMeans(n_clusters=k, n_init=5, random_state=42)
    labels = km.fit_predict(X)
    wcss.append(km.inertia_)
    silhouette_kmeans.append(silhouette_score(X, labels, sample_size=5000, random_state=42))
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=list(k_range), y=wcss, mode='lines+markers', name='Inertie (WCSS)'))
fig.update_layout(title="Méthode du coude — KMeans", xaxis_title='k', yaxis_title='Inertie', height=400, width=650)
fig.show()

fig = go.Figure()
fig.add_trace(go.Bar(x=list(k_range), y=silhouette_kmeans, name='Silhouette'))
fig.update_layout(title="Score de silhouette selon k", xaxis_title='k', yaxis_title='Silhouette', height=400, width=650)
fig.show()
```

```python
# Sélection automatique de k : meilleur silhouette parmi k in [3, 8]
# (on exclut k=2, trop grossier pour être exploitable opérationnellement)
candidats = [k for k in k_range if k >= 3]
scores_candidats = [silhouette_kmeans[k_range.index(k)] for k in candidats]
K_OPTIMAL = candidats[int(np.argmax(scores_candidats))]
print(f"k retenu = {K_OPTIMAL} (silhouette = {max(scores_candidats):.3f})")
```

```python
kmeans = KMeans(n_clusters=K_OPTIMAL, n_init=10, random_state=42)
sous_ensemble = sous_ensemble.copy()
sous_ensemble['Cluster_KMeans'] = kmeans.fit_predict(X)

fig = px.scatter_mapbox(
    sous_ensemble, lat='Lat', lon='Lon', color='Cluster_KMeans',
    mapbox_style='carto-positron', zoom=10, height=550,
    title=f"Clusters KMeans (k={K_OPTIMAL}) — {JOURS_FR[JOUR_PIC]} {HEURE_PIC}h",
    color_continuous_scale=px.colors.qualitative.Bold if K_OPTIMAL <= 10 else None,
)
fig.update_layout(margin=dict(l=0, r=0, t=40, b=0))
fig.show()
```

### 5.2 DBSCAN

DBSCAN doit calculer les distances entre points voisins, ce qui devient vite coûteux en mémoire quand `eps` est grand (chaque point peut avoir des milliers de voisins). Pour rester raisonnable en ressources tout en gardant un échantillon largement représentatif, on travaille ici sur un **sous-échantillon de 20 000 courses** (au lieu des 56 000+ courses de l'heure sélectionnée) — la même limite d'échantillonnage sera réutilisée en section 6 pour la généralisation à tous les jours.

```python
ECH_DBSCAN = min(20000, len(sous_ensemble))
sous_ensemble_db = sous_ensemble.sample(n=ECH_DBSCAN, random_state=42).copy()
X_db = sous_ensemble_db[['Lat', 'Lon']].to_numpy()
print(f"{len(sous_ensemble_db):,} courses utilisées pour DBSCAN".replace(',', ' '))
```

**Choix d'epsilon :** on utilise la méthode du "k-distance graph" : pour chaque point, on calcule la distance à son 2ᵉ plus proche voisin, puis on trie ces distances par ordre croissant. La courbe obtenue permet de lire à quelle distance la densité locale des points "casse".

```python
voisins = NearestNeighbors(n_neighbors=2).fit(X_db)
distances, _ = voisins.kneighbors(X_db)
distances = np.sort(distances[:, 1])

fig = px.line(y=distances, height=400, width=650,
              labels={'index': 'Points triés par distance croissante', 'y': 'Distance au 2ᵉ plus proche voisin'},
              title="k-distance graph")
fig.show()
```

On génère des valeurs candidates d'epsilon à partir des **percentiles élevés** de cette courbe (90ᵉ à 99.3ᵉ percentile) : cela couvre naturellement la zone du "coude" tout en évitant les toutes dernières valeurs extrêmes de la courbe, qui correspondent à quelques points très isolés et feraient exploser le temps/mémoire de calcul de DBSCAN sans améliorer la qualité des clusters. On compare ensuite score de silhouette (plus haut = mieux) et indice de Davies-Bouldin (plus bas = mieux), en ne retenant que les valeurs produisant une partition raisonnable (plus d'un cluster, peu de bruit).

```python
plage_eps = np.unique(np.round(np.percentile(distances, [90, 93, 95, 97, 98.5, 99.3]), 5))

resultats_eps = []
for eps in plage_eps:
    db = DBSCAN(eps=float(eps), min_samples=4, metric='euclidean').fit(X_db)
    labels = db.labels_
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    part_bruit = float((labels == -1).mean())
    if n_clusters < 2 or part_bruit > 0.4:
        continue
    sil = silhouette_score(X_db, labels, sample_size=5000, random_state=42)
    db_score = davies_bouldin_score(X_db, labels)
    resultats_eps.append((eps, n_clusters, part_bruit, sil, db_score))

resultats_eps_df = pd.DataFrame(
    resultats_eps, columns=['eps', 'n_clusters', 'part_bruit', 'silhouette', 'davies_bouldin']
)
resultats_eps_df
```

```python
import plotly.subplots as sp
fig = sp.make_subplots(rows=1, cols=2, subplot_titles=("Davies-Bouldin (plus bas = mieux)", "Silhouette (plus haut = mieux)"))
fig.add_trace(go.Scatter(x=resultats_eps_df['eps'], y=resultats_eps_df['davies_bouldin'], mode='lines+markers'), row=1, col=1)
fig.add_trace(go.Scatter(x=resultats_eps_df['eps'], y=resultats_eps_df['silhouette'], mode='lines+markers'), row=1, col=2)
fig.update_xaxes(title_text='eps', row=1, col=1)
fig.update_xaxes(title_text='eps', row=1, col=2)
fig.update_layout(height=400, width=900, showlegend=False, title_text="Sélection d'epsilon pour DBSCAN")
fig.show()
```

```python
# Sélection automatique : meilleur score de silhouette parmi les candidats valides
meilleure_ligne = resultats_eps_df.loc[resultats_eps_df['silhouette'].idxmax()]
EPS_OPTIMAL = float(meilleure_ligne['eps'])
print(f"eps retenu = {EPS_OPTIMAL:.5f} -> {int(meilleure_ligne['n_clusters'])} clusters, "
      f"{meilleure_ligne['part_bruit']*100:.1f}% de bruit, silhouette = {meilleure_ligne['silhouette']:.3f}")
```

```python
dbscan = DBSCAN(eps=EPS_OPTIMAL, min_samples=4, metric='euclidean')
sous_ensemble_db['Cluster_DBSCAN'] = dbscan.fit_predict(X_db)

fig = px.scatter_mapbox(
    sous_ensemble_db[sous_ensemble_db['Cluster_DBSCAN'] != -1],
    lat='Lat', lon='Lon', color='Cluster_DBSCAN',
    mapbox_style='carto-positron', zoom=10, height=550,
    title=f"Clusters DBSCAN (eps={EPS_OPTIMAL:.4f}) — {JOURS_FR[JOUR_PIC]} {HEURE_PIC}h — points de bruit exclus",
)
fig.update_layout(margin=dict(l=0, r=0, t=40, b=0))
fig.show()
```

### 5.3 Comparaison sur ce cas d'étude

```python
n_clusters_km = sous_ensemble['Cluster_KMeans'].nunique()
n_clusters_db = int(sous_ensemble_db.loc[sous_ensemble_db['Cluster_DBSCAN'] != -1, 'Cluster_DBSCAN'].nunique())
part_bruit_db = float((sous_ensemble_db['Cluster_DBSCAN'] == -1).mean())

sil_km = silhouette_score(X, sous_ensemble['Cluster_KMeans'], sample_size=5000, random_state=42)
sil_db = silhouette_score(X_db, sous_ensemble_db['Cluster_DBSCAN'], sample_size=5000, random_state=42)

print(f"KMeans  : {n_clusters_km} clusters (sur {len(sous_ensemble):,} courses), silhouette = {sil_km:.3f}".replace(',', ' '))
print(f"DBSCAN  : {n_clusters_db} clusters (sur {len(sous_ensemble_db):,} courses échantillonnées), "
      f"{part_bruit_db*100:.1f}% de points classés comme bruit, silhouette = {sil_db:.3f}".replace(',', ' '))
```

**Lecture :** KMeans impose un nombre de clusters fixe et produit des zones compactes et de tailles comparables — pratique pour donner une consigne simple aux chauffeurs ("va vers la zone 3"). DBSCAN détecte un nombre de clusters variable, épouse mieux la forme réelle de la densité de courses, et isole explicitement les courses isolées (bruit) qui ne correspondent à aucune vraie zone de forte demande — mais il est plus sensible au réglage de `eps` et plus coûteux en mémoire sur de gros volumes.

## 6. Généralisation : une carte de zones chaudes par jour de la semaine

On applique maintenant la même méthode à **chaque jour de la semaine**, du **lundi** au **dimanche**, en utilisant l'ensemble des courses de ce jour cumulées sur les 6 mois (et non plus une seule heure), afin d'obtenir une vue journalière des zones à forte demande.

```python
import gc

dfs_par_jour = {
    jour: df.loc[df['day_of_week'] == jour, ['Lat', 'Lon', 'hour', 'jour_fr']].copy()
    for jour in JOURS_ORDRE
}

for jour in JOURS_ORDRE:
    print(f"{JOURS_FR[jour]:>10} : {len(dfs_par_jour[jour]):>9,} courses".replace(',', ' '))
```

### 6.1 KMeans par jour

On garde `k=5` pour chaque jour : cela correspond, dans notre cas d'étude ci-dessus, à un bon compromis entre score de silhouette et lisibilité opérationnelle (5 zones = 5 consignes simples à donner à des chauffeurs), et permet de comparer les jours entre eux sur une base commune.

Comme chaque jour représente ici plusieurs centaines de milliers de courses, on utilise **`MiniBatchKMeans`** plutôt que `KMeans` classique : c'est une variante conçue pour passer à l'échelle sur de gros volumes (elle travaille par petits lots plutôt que sur tout le jeu de données à chaque itération), pour un résultat très proche en pratique.

```python
from sklearn.cluster import MiniBatchKMeans

K_JOURNALIER = 5

for jour in JOURS_ORDRE:
    X_jour = dfs_par_jour[jour][['Lat', 'Lon']].to_numpy()
    kmeans = MiniBatchKMeans(n_clusters=K_JOURNALIER, random_state=42, n_init=5, batch_size=2048)
    dfs_par_jour[jour]['Cluster_KMeans'] = kmeans.fit_predict(X_jour)
    print(f"{JOURS_FR[jour]:>10} : clustering KMeans terminé")

_ = gc.collect()
```

```python
def carte_kmeans(jour, echantillon_max=15000):
    d = dfs_par_jour[jour]
    if len(d) > echantillon_max:
        d = d.sample(n=echantillon_max, random_state=42)
    fig = px.scatter_mapbox(
        d, lat='Lat', lon='Lon', color='Cluster_KMeans',
        mapbox_style='carto-positron', zoom=9.5, height=450,
        title=f"Zones chaudes KMeans (k={K_JOURNALIER}) — {JOURS_FR[jour]}",
    )
    fig.update_layout(margin=dict(l=0, r=0, t=40, b=0))
    return fig

for jour in JOURS_ORDRE:
    carte_kmeans(jour).show()
```

### 6.2 DBSCAN par jour

DBSCAN sur un jour complet représente plusieurs centaines de milliers de points : la matrice de distances devient trop coûteuse en mémoire. On échantillonne donc à 20 000 points par jour avant d'appliquer l'algorithme — ce qui reste largement représentatif de la structure géographique de la demande.

```python
EPS_JOURNALIER = EPS_OPTIMAL  # même epsilon que sur le cas d'étude, cohérent car même échelle géographique (NYC)
TAILLE_ECHANTILLON = 20000

for jour in JOURS_ORDRE:
    d = dfs_par_jour[jour]
    if len(d) > TAILLE_ECHANTILLON:
        d = d.sample(n=TAILLE_ECHANTILLON, random_state=42)
        dfs_par_jour[jour] = d.copy()  # on retient l'échantillon utilisé, pour rester cohérent avec le clustering
    X_jour = d[['Lat', 'Lon']].to_numpy()
    dbscan = DBSCAN(eps=EPS_JOURNALIER, min_samples=4, metric='euclidean')
    dfs_par_jour[jour]['Cluster_DBSCAN'] = dbscan.fit_predict(X_jour)

_ = gc.collect()
print("Clustering DBSCAN terminé pour les 7 jours (échantillon de 20 000 courses par jour).")
```

```python
def carte_dbscan(jour):
    d = dfs_par_jour[jour]
    d = d[d['Cluster_DBSCAN'] != -1]
    fig = px.scatter_mapbox(
        d, lat='Lat', lon='Lon', color='Cluster_DBSCAN',
        mapbox_style='carto-positron', zoom=9.5, height=450,
        title=f"Zones chaudes DBSCAN (eps={EPS_JOURNALIER:.4f}) — {JOURS_FR[jour]} — bruit exclu",
    )
    fig.update_layout(margin=dict(l=0, r=0, t=40, b=0))
    return fig

for jour in JOURS_ORDRE:
    carte_dbscan(jour).show()
```

### 6.3 Synthèse chiffrée par jour

```python
lignes = []
for jour in JOURS_ORDRE:
    d = dfs_par_jour[jour]
    n_km = d['Cluster_KMeans'].nunique()
    sil_km = silhouette_score(d[['Lat', 'Lon']], d['Cluster_KMeans'], sample_size=5000, random_state=42)
    n_db = int((d['Cluster_DBSCAN'].unique() != -1).sum())
    part_bruit = float((d['Cluster_DBSCAN'] == -1).mean())
    sil_db = silhouette_score(d[['Lat', 'Lon']], d['Cluster_DBSCAN'], sample_size=5000, random_state=42)
    lignes.append({
        'Jour': JOURS_FR[jour],
        'Courses (échantillon DBSCAN)': len(d),
        'Clusters KMeans': n_km,
        'Silhouette KMeans': round(sil_km, 3),
        'Clusters DBSCAN': n_db,
        '% bruit DBSCAN': round(part_bruit * 100, 1),
        'Silhouette DBSCAN': round(sil_db, 3),
    })

synthese = pd.DataFrame(lignes)
synthese
```

On assemble les résultats KMeans des 7 jours dans une seule carte interactive, avec un curseur pour naviguer du **lundi** au **dimanche** — c'est la vue destinée à être présentée aux équipes Uber.

## 8. Extension : zones nommées avec le dataset 2015

Le jeu de données `uber-raw-data-janjune-15` (janvier à juin 2015) ne contient pas de coordonnées GPS : chaque course est associée à un `locationID`, un identifiant renvoyant vers l'une des 263 zones officielles définies par la ville de New York (fichier `taxi-zone-lookup.csv`, avec le nom du quartier — `Borough` — et de la zone — `Zone`).

On ne peut donc pas y appliquer KMeans ou DBSCAN de la même façon (il n'y a pas de coordonnées continues à regrouper), mais ce jeu de données est complémentaire du précédent : là où nos clusters GPS donnent des zones géométriques sans nom, celui-ci donne directement des **zones nommées et reconnaissables** (ex. "Midtown Center", "Union Sq"), ce qui est très utile pour présenter des résultats lisibles à des équipes non techniques. Avec 14+ millions de lignes, on agrège les données par lots (`chunksize`) pour rester léger en mémoire, sans avoir besoin de charger le fichier en entier.

```python
import zipfile

lookup = pd.read_csv('taxi-zone-lookup.csv')[['LocationID', 'Borough', 'Zone']]

agg_total = None
nb_lignes_2015 = 0

with zipfile.ZipFile('uber-raw-data-janjune-15.zip') as z:
    with z.open('uber-raw-data-janjune-15.csv') as f:
        lecteur = pd.read_csv(f, usecols=['Pickup_date', 'locationID'], chunksize=2_000_000)
        for morceau in lecteur:
            nb_lignes_2015 += len(morceau)
            dt = pd.to_datetime(morceau['Pickup_date'], format='%Y-%m-%d %H:%M:%S')
            morceau['day_of_week'] = dt.dt.day_name()
            morceau['hour'] = dt.dt.hour
            g = morceau.groupby(['day_of_week', 'hour', 'locationID'], observed=True).size()
            agg_total = g if agg_total is None else agg_total.add(g, fill_value=0)

agg_2015 = agg_total.rename('nb_courses').reset_index()
agg_2015['day_of_week'] = pd.Categorical(agg_2015['day_of_week'], categories=JOURS_ORDRE, ordered=True)
agg_2015['jour_fr'] = agg_2015['day_of_week'].map(JOURS_FR)
agg_2015['jour_fr'] = pd.Categorical(agg_2015['jour_fr'], categories=[JOURS_FR[j] for j in JOURS_ORDRE], ordered=True)
agg_2015 = agg_2015.merge(lookup, left_on='locationID', right_on='LocationID', how='left')

print(f"{nb_lignes_2015:,} courses agrégées (jan.-juin 2015)".replace(',', ' '))
agg_2015.head()
```

### 8.1 Zones les plus demandées, tous jours confondus

```python
top_zones = (
    agg_2015.groupby(['Borough', 'Zone'], observed=True)['nb_courses']
    .sum()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)
top_zones['Zone_complete'] = top_zones['Zone'] + ' (' + top_zones['Borough'] + ')'

fig = px.bar(
    top_zones.sort_values('nb_courses'),
    x='nb_courses', y='Zone_complete', orientation='h',
    labels={'nb_courses': 'Nombre de courses (jan.-juin 2015)', 'Zone_complete': ''},
    title="Top 15 des zones New York les plus demandées",
    height=500,
)
fig.show()
```

### 8.2 Zone la plus demandée, jour par jour (lundi → dimanche)

```python
top_par_jour = (
    agg_2015.groupby(['jour_fr', 'Borough', 'Zone'], observed=True)['nb_courses']
    .sum()
    .reset_index()
    .sort_values('nb_courses', ascending=False)
    .groupby('jour_fr', observed=True)
    .head(1)
    .sort_values('jour_fr')
    [['jour_fr', 'Zone', 'Borough', 'nb_courses']]
    .rename(columns={'jour_fr': 'Jour'})
)
top_par_jour
```

**Lecture :** cette table confirme, avec un jeu de données différent et une granularité différente (zones officielles plutôt que clusters GPS), la même intuition que notre clustering en section 6 : ce sont des zones centrales et bien desservies (Manhattan en tête) qui concentrent le plus de demande, tous les jours de la semaine — une bonne validation croisée de l'approche.

## 9. Conclusion

**Ce qu'on retient de la comparaison KMeans / DBSCAN sur ce projet :**

- **KMeans** produit un nombre de zones fixe et stable d'un jour à l'autre, ce qui facilite la comparaison et donne des consignes simples et actionnables pour les chauffeurs. C'est l'algorithme le plus adapté pour une recommandation opérationnelle du type "positionnez-vous en zone X".
- **DBSCAN** s'adapte à la densité réelle des courses et isole naturellement les courses isolées (bruit), ce qui est utile pour la partie diagnostic/exploration, mais il est plus sensible au réglage de ses hyperparamètres (`eps`, `min_samples`) et plus coûteux à faire tourner sur de gros volumes (d'où l'échantillonnage nécessaire).
- Le jeu de données 2015 par zones nommées (section 8) confirme, indépendamment, les mêmes grandes tendances géographiques que notre clustering GPS — une bonne validation croisée de la méthode.

**Recommandation :** proposer à Uber un déploiement basé sur **KMeans** (k=5 par jour) pour la recommandation en temps réel aux chauffeurs, avec un affichage utilisant les noms de zones officielles (via un rapprochement géographique cluster ↔ zone TLC) plutôt que de simples coordonnées, pour rester lisible côté chauffeur. **DBSCAN** reste un outil précieux d'analyse exploratoire pour affiner la définition des zones (fusion/scission selon les jours).

**Limites et pistes d'amélioration :**
- Le clustering est ici uniquement spatial ; il pourrait être affiné en travaillant heure par heure (comme fait en section 5) plutôt que jour entier, pour capter les variations intra-journée (heures de pointe du matin vs du soir).
- Un découpage horaire glissant (par ex. par tranche de 2h) donnerait des recommandations encore plus proches du besoin réel des chauffeurs en temps réel.
- Une jointure géométrique propre entre les clusters GPS (section 6) et les zones officielles TLC (section 8) — via les polygones de zones, non fournis ici — permettrait d'unifier les deux approches en une seule vue nommée et cartographiée.
